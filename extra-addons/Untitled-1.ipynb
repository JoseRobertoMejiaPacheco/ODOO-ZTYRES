```python
import sys
import pandas as pd
from datetime import datetime
import re
sys.path.append('/usr/lib/python3/dist-packages/odoo')  # Ruta base de Odoo
sys.path.append('/mnt/extra-addons')  # Tu ruta de addons personalizados
import pytz
import odoo
import odoo.modules.registry
from odoo import api, tools
local_tz = pytz.timezone('America/Mexico_City')

config = tools.config
config['db_name'] = 'ZTYRES'
config['addons_path'] = '/mnt/extra-addons,/usr/lib/python3/dist-packages/odoo/addons'
# config['log_level'] = 'info'
config['config_file'] = '/etc/odoo/odoo.conf'

db_name = config['db_name']
registry = odoo.modules.registry.Registry(db_name)
cr = registry.cursor()
env = api.Environment(cr, odoo.SUPERUSER_ID, {})
print(1)
```

```python
domain = [('name','in',[
'VNAC/2025/10917',
'VNAC/2025/10953',
'VNAC/2025/11112',
'VNAC/2025/11325',
'VNAC/2025/11498',
'VNAC/2025/11716',
'VNAC/2025/11816',
'VNAC/2025/11993',
'VNAC/2025/11992',
'VNAC/2025/12079',
'VNAC/2025/12251',
'VNAC/2025/12372',
'VNAC/2025/12590',
'VEMP/2025/00001',
'VEMP/2025/00002',
'VEMP/2025/00004',
'VEMP/2025/00003',
'VEMP/2025/00005',
'VEMP/2025/00006',
'VNAC/2025/14549',
'VNAC/2025/14550',
'VNAC/2025/14735',
'VEMP/2025/00007',
'VNAC/2025/14863',
'VNAC/2025/15132'
])]
data = []
invoices = env['account.move'].search(domain)
lines = invoices.invoice_line_ids
for line in lines:    
    matches = re.findall(r'<strong.*?>(.*?)</strong>', str(line.sale_line_ids.order_id.note))
    # Elimina cualquier etiqueta HTML interna
    matches = [re.sub(r'<.*?>', '', m) for m in matches]
    print(line.move_id.name,line.move_id.invoice_date,matches,line.move_id.amount_total)
    data.append({
        'Factura': line.move_id.name,
        'Fecha': line.move_id.invoice_date,
        'Nota Strong': matches[0] if matches else '',
        'Total': line.move_id.amount_total
    })

# Crear DataFrame
df = pd.DataFrame(data)

# Exportar a Excel
df.to_excel('Venta Empleados.xlsx', index=False)

```
